Raw text contains noise (punctuation, numbers, symbols, stopwords)

Convert text labels into numeric labels

Convert cleaned text into TF-IDF vectors

🔹 Import Libraries & Download Resources

```python
import nltk                  # Natural Language Toolkit for NLP tasks
import re                    # Regular expressions for text cleaning
import pandas as pd          # Used to store and save processed data

# Import NLP tools from NLTK
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Import ML tools from sklearn
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download('punkt')       # Tokenizer models
nltk.download('stopwords')   # Stopword list
nltk.download('wordnet')     # Lemmatization dictionary
nltk.download('omw-1.4')     # WordNet extra data
```

```python
texts = [
    "I absolutely loved the movie!!! 😍",
    "The movie was boring and too long...",
    "What a fantastic performance by the actors!",
    "I did not like the film at all 😡",
    "The story was average, nothing special."
]

labels = ["positive", "negative", "positive", "negative", "neutral"]
```

🔹 TEXT CLEANING

What we do: Lowercase, Remove punctuation, Remove numbers
, Remove emojis & special characters



```python
def clean_text(text):
    """
    This function cleans the input text by:
    - Converting to lowercase
    - Removing punctuation, numbers, emojis
    """
    text = text.lower()                       # Convert to lowercase
    text = re.sub(r'[^a-z\s]', '', text)      # Remove punctuation, numbers, emojis
    return text
```

```python
cleaned_texts = [clean_text(text) for text in texts]

print("Cleaned Text:")
for t in cleaned_texts:
    print(t)
```

🔹 STOP WORD REMOVAL

Stop words:
Common words like is, the, was, and that carry little meaning

```python
# Load English stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    """
    Removes stopwords from the text
    """
    words = word_tokenize(text)
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)
```

```python
import nltk
nltk.download('punkt_tab') # Download the missing resource

no_stopwords_texts = [remove_stopwords(text) for text in cleaned_texts]

print("\nText after Stopword Removal:")
for t in no_stopwords_texts:
    print(t)
```

🔹 LEMMATIZATION (WordNet)


Why Lemmatization?

Converts words to dictionary base form

Example: loved → love, actors → actor

```python
lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    """
    Lemmatizes each word in the text
    """
    words = word_tokenize(text)
    lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
    return " ".join(lemmatized_words)
```

```python
lemmatized_texts = [lemmatize_text(text) for text in no_stopwords_texts]

print("\nLemmatized Text:")
for t in lemmatized_texts:
    print(t)
```

🔹 LABEL ENCODING

Purpose:
Convert categorical labels → numerical form

```python
# Create LabelEncoder object
label_encoder = LabelEncoder()

# Encode labels
encoded_labels = label_encoder.fit_transform(labels)

print("\nOriginal Labels:", labels)
print("Encoded Labels:", encoded_labels)
```

🔹 TF-IDF REPRESENTATION

```python
# Create TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit and transform lemmatized text
tfidf_matrix = tfidf_vectorizer.fit_transform(lemmatized_texts)

# Convert to array
tfidf_array = tfidf_matrix.toarray()

print("\nTF-IDF Feature Names:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_array)
```

🔹 9. SAVE OUTPUTS (IMPORTANT 🔥)

We save:
Cleaned text,
Encoded labels,
TF-IDF matrix

```python
# Create DataFrame
df = pd.DataFrame(tfidf_array, columns=tfidf_vectorizer.get_feature_names_out())
df['Label'] = encoded_labels
df['Processed_Text'] = lemmatized_texts

# Save to CSV file
df.to_csv("processed_text_tfidf_output.csv", index=False)

print("\nOutput saved as 'processed_text_tfidf_output.csv'")
```
